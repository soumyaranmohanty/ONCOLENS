Need to check further for the skewness.

```python
#convert HISTORY_NEOADJUVANT_TRTYN to binary
selected_clinical["HISTORY_NEOADJUVANT_TRTYN"] = selected_clinical["HISTORY_NEOADJUVANT_TRTYN"].map({"Yes": 1, "No": 0})
```

```python
#convert NEW_TUMOR_EVENT_AFTER_INITIAL_TREATMENT, RADIATION_THERAPY  to binary
selected_clinical["NEW_TUMOR_EVENT_AFTER_INITIAL_TREATMENT"] = selected_clinical["NEW_TUMOR_EVENT_AFTER_INITIAL_TREATMENT"].map({"Yes": 1, "No": 0})
selected_clinical["RADIATION_THERAPY"] = selected_clinical["RADIATION_THERAPY"].map({"Yes": 1, "No": 0})
```

```python
#convert 0:LIVING, 1:DECEASED to 0 and 1
selected_clinical["OS_STATUS"] = selected_clinical["OS_STATUS"].map({"0:LIVING": 0, "1:DECEASED": 1})
```

```python
#convert 0:CENSORED, 1:DECEASED to 0 and 1
selected_clinical["PFS_STATUS"] = selected_clinical["PFS_STATUS"].map({"0:CENSORED": 0, "1:PROGRESSION": 1})
```

```python
#Convert PRIOR_DX to binary
selected_clinical["PRIOR_DX"] = selected_clinical["PRIOR_DX"].map({"No": 0, "Yes": 1, "Yes, History Of Synchronous And Or Bilateral Malignancy": 1})
```

```python
selected_clinical["AJCC_PATHOLOGIC_TUMOR_STAGE"] = selected_clinical["AJCC_PATHOLOGIC_TUMOR_STAGE"].map(
{
 'STAGE I':1.0,
 'STAGE IA':1.1,
 'STAGE IB':1.2,
 'STAGE II':2.0,
 'STAGE IIA':2.1,
 'STAGE IIB':2.2,
 'STAGE IIIA':3.1,
 'STAGE IIIB':3.2,
'STAGE IV':4.0
})

selected_clinical["PATH_T_STAGE"] = selected_clinical["PATH_T_STAGE"].map(
{
 'T1':1.0,
 'T1A':1.1,
 'T1B':1.2,
 'T2':2.0,
 'T2A':2.1,
 'T2B':2.2,
 'T3':3.0,
 'T4':4.0,
 'TX':-1
})

selected_clinical["PATH_N_STAGE"] = selected_clinical["PATH_N_STAGE"].map(
{
 'N0':0,
 'N1':1,
 'N2':2,
 'N3':3,
 'NX':-1
})

selected_clinical["PATH_M_STAGE"] = selected_clinical["PATH_M_STAGE"].map(
{
 'M0':0,
 'M1':1,
 'M1A':1.1,
 'M1B':1.2,
 'MX':-1
})
```
